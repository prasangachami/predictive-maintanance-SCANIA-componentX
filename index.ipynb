```python
# Load Libreries
import numpy as np
import pandas as pd
import matplotlib as plt
import os 
```

```python
# Load all the files that we want to load from the path of the data folder
data_path = "data"

# List all the .csv files
files = [f for f in os.listdir(data_path) if f.endswith(".csv")]
print("files count in the data folder", len(files))
for file in files:
    print(f"\n Loading : {file}")
    df = pd.read_csv(os.path.join(data_path, file))

    # Shape of the dataset
    print("Shape ", df.shape)

    # Column names - features
    print("Columns list", df.columns.tolist())

    # Preview first rows
    print(df.head())

    # Preview last rows
    print(df.tail())
```
